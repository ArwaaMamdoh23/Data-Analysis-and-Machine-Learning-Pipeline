```python
import threading
import random


ARRAY_SIZE = 100
NUM_THREADS = 4

arr = [random.randint(0, 9) for _ in range(ARRAY_SIZE)]
output_lock = threading.Lock()

#  search for a specific number in the array
def search_in_array(start, end, number):
    found_indices = []
    for i in range(start, end):
        if arr[i] == number:
            found_indices.append(i)


    with output_lock:
        if found_indices:
            print(f"Number {number} found at indices: {found_indices} by thread ID: {threading.get_ident()}")

def main():
    number_to_search = 4
    threads = []


    segment_size = ARRAY_SIZE // NUM_THREADS
    for i in range(NUM_THREADS):
        start = i * segment_size
        end = (i + 1) * segment_size if i < NUM_THREADS - 1 else ARRAY_SIZE
        thread = threading.Thread(target=search_in_array, args=(start, end, number_to_search))
        threads.append(thread)
        thread.start()

    # Join the threads
    for thread in threads:
        thread.join()

if __name__ == "__main__":
    main()
```
